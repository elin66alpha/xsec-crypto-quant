# Phase 0 Demo 02: Dynamic Universe Evolution

This notebook demonstrates monthly dynamic-universe selection on a small liquid OKX
perpetual subset. It is a smoke demo, not the final full Top N research universe.

```python
import pandas as pd
import plotly.express as px

from data.universe import UniverseConfig, build_universe_history, load_dollar_volume_panel
```

```python
symbols = [
    "BTC/USDT:USDT",
    "ETH/USDT:USDT",
    "SOL/USDT:USDT",
    "XRP/USDT:USDT",
    "DOGE/USDT:USDT",
]

until = pd.Timestamp.now("UTC").normalize()
since = until - pd.Timedelta(days=150)
panel = load_dollar_volume_panel(symbols=symbols, since=since, until=until)
assert not panel.empty, "dollar volume panel is empty"

config = UniverseConfig(top_n=3, min_listing_days=30, lookback_days=30)
history = build_universe_history(
    panel,
    start=until - pd.DateOffset(months=2),
    end=until,
    config=config,
)
history
```

```python
rows = []
for month_start, members in history.items():
    for rank, symbol in enumerate(members, start=1):
        rows.append({"month": month_start, "rank": rank, "symbol": symbol})

ranks = pd.DataFrame(rows)
ranks
```

```python
fig = px.line(
    ranks,
    x="month",
    y="rank",
    color="symbol",
    markers=True,
    title="Recent OKX Demo Universe Ranks",
)
fig.update_yaxes(autorange="reversed", dtick=1)
fig.show()
```
